### 2. Load Model and Tokenizer

```python
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("openai-community/gpt2")
model = AutoModelForCausalLM.from_pretrained("openai-community/gpt2")
```

### 3. Implement a Filtering Mechanism

```python
def is_python_question(prompt):
    python_keywords = [
        "python", "code", "function", "loop", "list", "dictionary",
        "tuple", "numpy", "pandas", "class", "object", "def",
        "lambda", "exception", "error", "debug", "script"
    ]

    prompt_lower = prompt.lower()
    return any(keyword in prompt_lower for keyword in python_keywords)
```

### 4. Generate Response

```python
def generate_python_response(prompt):
    # Add instruction to constrain output
    instruction = (
        "You are a Python coding assistant. Answer only Python-related questions.\n"
        f"Question: {prompt}\nAnswer:"
    )

    inputs = tokenizer.encode(instruction, return_tensors="pt")

    outputs = model.generate(
        inputs,
        max_length=150,
        num_return_sequences=1,
        temperature=0.7,
        top_p=0.9,
        do_sample=True,
        pad_token_id=tokenizer.eos_token_id
    )

    response = tokenizer.decode(outputs[0], skip_special_tokens=True)

    # Extract only the generated answer part
    return response.split("Answer:")[-1].strip()
```

### 5. Handle Non-Coding Questions

```python
def code_focused_inference(prompt):
    if is_python_question(prompt):
        return generate_python_response(prompt)
    else:
        return "Sorry, I can only answer Python coding-related questions."
```

### 6. Test

```python
test_prompts = [
    "How to write a for loop in Python?",
    "Explain Python list comprehension",
    "What is the capital of France?",
    "Write a Python function to reverse a string",
    "Tell me a joke"
]

for i, prompt in enumerate(test_prompts, 1):
    print(f"\nTest Case {i}: {prompt}")
    print("Response:", code_focused_inference(prompt))
```
