## Data Preparation

### Subtask:
Load the 'data.csv' dataset, convert the 'Date' column to datetime objects, sort the data by date, and create new features: 'Year', 'Month', 'Day_of_Month', 'Day_of_Week', 'Weekday_Name', 'Daily_Return', and 'Next_Day_Return'. Explain the meaning of each newly created feature.


**Reasoning**:
The subtask requires loading a CSV, converting and sorting dates, creating several new date and return-based features, and then displaying the head and info of the DataFrame. These steps can be efficiently performed in a single code block.



````python
import pandas as pd

# Load the 'data.csv' file
df = pd.read_csv('/content/data.csv')

# Convert the 'Date' column to datetime objects
df['Date'] = pd.to_datetime(df['Date'])

# Sort the DataFrame by the 'Date' column and reset index
df = df.sort_values(by='Date').reset_index(drop=True)

# Create new columns derived from the 'Date' column
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day_of_Month'] = df['Date'].dt.day
df['Day_of_Week'] = df['Date'].dt.dayofweek
df['Weekday_Name'] = df['Date'].dt.day_name()

# Calculate 'Daily_Return' as the percentage change of the 'Close' price
df['Daily_Return'] = df['Close'].pct_change() * 100

# Calculate 'Next_Day_Return' as the 'Daily_Return' shifted by one day
df['Next_Day_Return'] = df['Daily_Return'].shift(-1)

# Print the first few rows of the DataFrame
print("DataFrame head after feature creation:")
print(df.head())

# Print the data types of all columns
print("\nDataFrame info after feature creation:")
df.info()
````

### Explanation of Newly Created Features:

*   **Year**: Represents the calendar year of the trading day, extracted from the original `Date` column.
*   **Month**: Represents the numerical month of the trading day, where 1 signifies January and 12 signifies December.
*   **Day_of_Month**: Indicates the specific day of the month for each trading entry (e.g., the 1st, 15th, or 30th).
*   **Day_of_Week**: A numerical representation of the day of the week, with Monday typically being 0 and Sunday being 6.
*   **Weekday_Name**: The full name of the day of the week (e.g., 'Monday', 'Tuesday'), providing a more human-readable format.
*   **Daily_Return**: The percentage change in the `Close` price from the previous trading day. A positive value indicates a gain, while a negative value indicates a loss for that specific day.
*   **Next_Day_Return**: The percentage change in the `Close` price that occurs on the day *following* the current date. This feature is particularly useful for evaluating investment strategies, as an investment made on a given day would realize the return of the next trading day.

**Reasoning**:
I need to define a function to simulate SIP investments for a given day of the month and then calculate the cumulative growth for the specified days (Day 1, Day 5, Day 10, Day 15, Day 20, and Day 25) as required by the subtask.



````python
investment_days = [1, 5, 10, 15, 20, 25]
sip_results = {}

# Function to simulate SIP for a given day of the month
def simulate_sip_growth(data, investment_day, initial_investment=1000):
    # Filter data for the specific investment day
    sip_df = data[data['Day_of_Month'] == investment_day].copy()

    # Ensure there's data for the investment day
    if sip_df.empty:
        return pd.Series([1], index=data['Date'].min())

    # Calculate the daily growth factor (1 + Next_Day_Return / 100)
    # Assuming 'Next_Day_Return' is already in percentage
    sip_df['Growth_Factor'] = 1 + sip_df['Next_Day_Return'] / 100

    # Drop rows with NaN in Growth_Factor, which would occur if Next_Day_Return is NaN
    sip_df.dropna(subset=['Growth_Factor'], inplace=True)

    # Start with an initial investment for the first investment period
    cumulative_value = [initial_investment]

    # Simulate SIP: Each investment adds `initial_investment` to the portfolio,
    # and the entire portfolio grows by the `Growth_Factor` of the current investment day.

    # For the first day, we just invest the initial amount.
    # For subsequent investments, the new investment is added to the previous cumulative value
    # and then multiplied by the growth factor for that period.

    # Align investment dates with actual trading days
    sip_dates = sip_df['Date'].values

    # Create a Series for cumulative growth indexed by Date
    # Initialize with the first investment
    cumulative_growth_series = pd.Series(dtype='float64')
    if not sip_df.empty:
        cumulative_growth_series.loc[sip_df['Date'].iloc[0]] = initial_investment

    # Track the total number of investments made
    total_investments = 1

    for i in range(1, len(sip_df)):
        current_growth_factor = sip_df['Growth_Factor'].iloc[i]
        current_date = sip_df['Date'].iloc[i]

        # The portfolio value from the previous period (before new investment) grows
        previous_portfolio_value = cumulative_value[-1]
        grown_previous_portfolio_value = previous_portfolio_value * current_growth_factor

        # Add new investment and update cumulative value
        new_cumulative_value = grown_previous_portfolio_value + initial_investment
        cumulative_value.append(new_cumulative_value)
        cumulative_growth_series.loc[current_date] = new_cumulative_value
        total_investments += 1

    # Adjust cumulative value to reflect only the growth of initial investments if needed
    # For a true SIP return, we should normalize by total investment amount

    if not cumulative_growth_series.empty:
        # Ensure the series is sorted by date
        cumulative_growth_series = cumulative_growth_series.sort_index()

        # Normalize the growth: divide by total invested capital
        # To get a growth multiplier, we need to divide by the total amount invested.
        # The `cumulative_value` list contains the total portfolio value at each investment point.
        # We need to create a Series of total invested amounts corresponding to these points.
        total_invested_at_each_step = pd.Series([initial_investment * (j + 1) for j in range(total_investments)], index=cumulative_growth_series.index)

        # Calculate the growth multiplier (Current Value / Total Invested)
        # For visualization of 'growth', we can just plot the raw cumulative value
        # or normalize it. The task asks for 'cumulative SIP growth'.
        # Let's keep it as the total value for now, as it represents growth of capital.
        return cumulative_growth_series
    else:
        return pd.Series(dtype='float64')


# Calculate cumulative SIP growth for selected days
for day in investment_days:
    sip_results[f'Day_{day}_SIP_Growth'] = simulate_sip_growth(df, day)

# Convert results to a DataFrame for easier plotting
sip_growth_df = pd.DataFrame({k: v for k, v in sip_results.items() if not v.empty})

print("Cumulative SIP growth calculated for selected days:")
print(sip_growth_df.head())
print(sip_growth_df.tail())
````

The subtask requires visualizing the cumulative SIP growth for the selected days (1, 5, 10, 15, 20, 25). A single interactive Plotly line chart will be used to show these trends, adhering to the specified dark theme and interactive elements.



````python
import plotly.graph_objects as go

# Visualize cumulative SIP growth for selected days
fig_sip_growth = px.line(
    sip_growth_df.reset_index(),
    x='index',
    y=sip_growth_df.columns,
    title='Cumulative SIP Growth for Different Investment Days of the Month',
    template='plotly_dark',
    labels={'index': 'Date', 'value': 'Cumulative Value', 'variable': 'Investment Day'}
)

fig_sip_growth.update_layout(
    xaxis_title='Date',
    yaxis_title='Cumulative Value ($)',
    hovermode='x unified'
)
fig_sip_growth.show()
````

### Subtask:
Simulate and compare cumulative investment growth if SIP occurs every Monday, Tuesday, Wednesday, Thursday, and Friday.


**Reasoning**:
I need to define a flexible SIP simulation function that can filter by any given column (e.g., 'Day_of_Month' or 'Weekday_Name') and then use this function to calculate the cumulative SIP growth for each weekday, storing the results. Finally, I will display the head and tail of the results.



````python
investment_weekdays = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday']
sip_weekday_results = {}

# Flexible function to simulate SIP for a given filter (e.g., day of month, weekday)
def simulate_sip_growth_flexible(data, filter_column_name, filter_value, initial_investment=1000):
    # Filter data for the specific investment period
    sip_df = data[data[filter_column_name] == filter_value].copy()

    # Ensure there's data for the investment period
    if sip_df.empty:
        return pd.Series(dtype='float64')

    # Calculate the daily growth factor (1 + Next_Day_Return / 100)
    # Assuming 'Next_Day_Return' is already in percentage
    sip_df['Growth_Factor'] = 1 + sip_df['Next_Day_Return'] / 100

    # Drop rows with NaN in Growth_Factor, which would occur if Next_Day_Return is NaN
    sip_df.dropna(subset=['Growth_Factor'], inplace=True)

    if sip_df.empty: # Check again after dropping NaNs
        return pd.Series(dtype='float64')

    # Initialize cumulative value and series
    cumulative_value = [initial_investment]
    cumulative_growth_series = pd.Series(dtype='float64')
    cumulative_growth_series.loc[sip_df['Date'].iloc[0]] = initial_investment

    total_investments = 1

    for i in range(1, len(sip_df)):
        current_growth_factor = sip_df['Growth_Factor'].iloc[i]
        current_date = sip_df['Date'].iloc[i]

        previous_portfolio_value = cumulative_value[-1]
        grown_previous_portfolio_value = previous_portfolio_value * current_growth_factor

        new_cumulative_value = grown_previous_portfolio_value + initial_investment
        cumulative_value.append(new_cumulative_value)
        cumulative_growth_series.loc[current_date] = new_cumulative_value
        total_investments += 1

    if not cumulative_growth_series.empty:
        # Ensure the series is sorted by date
        cumulative_growth_series = cumulative_growth_series.sort_index()
        return cumulative_growth_series
    else:
        return pd.Series(dtype='float64')

# Calculate cumulative SIP growth for selected weekdays
for day_name in investment_weekdays:
    sip_weekday_results[f'{day_name}_SIP_Growth'] = simulate_sip_growth_flexible(df, 'Weekday_Name', day_name)

# Convert results to a DataFrame for easier plotting
sip_growth_weekday_df = pd.DataFrame({k: v for k, v in sip_weekday_results.items() if not v.empty})

print("Cumulative SIP growth calculated for selected weekdays:")
print(sip_growth_weekday_df.head())
print(sip_growth_weekday_df.tail())
````

## Yearly Patterns Analysis

### Subtask:
Examine yearly market behavior, including bull and bear periods. Compute yearly returns, volatility, and maximum drawdown.


**Reasoning**:
I need to calculate yearly returns, yearly volatility, and maximum drawdown for each year, then consolidate these metrics into a DataFrame, and display its head and tail as per the instructions.



````python
import numpy as np

# Helper function to calculate Max Drawdown
def calculate_max_drawdown(prices):
    if len(prices) < 2:
        return 0
    peak = prices[0]
    max_drawdown = 0
    for price in prices:
        if price > peak:
            peak = price
        drawdown = (peak - price) / peak * 100
        if drawdown > max_drawdown:
            max_drawdown = drawdown
    return max_drawdown

# Group by year
yearly_groups = df.groupby('Year')

yearly_data = []

# Calculate yearly metrics
for year, group in yearly_groups:
    # Yearly Return
    # Get the last close price of the current year
    current_year_last_close = group['Close'].iloc[-1]

    # Get the last close price of the previous year
    previous_year_last_close = df[df['Year'] == year - 1]['Close'].iloc[-1] if year > df['Year'].min() else np.nan

    # Calculate yearly return, handling the first year
    if not np.isnan(previous_year_last_close):
        yearly_return = (current_year_last_close / previous_year_last_close - 1) * 100
    else:
        yearly_return = np.nan # No previous year to compare

    # Yearly Volatility (standard deviation of daily returns for that year)
    yearly_volatility = group['Daily_Return'].std()

    # Max Drawdown for the year
    max_drawdown = calculate_max_drawdown(group['Close'].values)

    yearly_data.append({
        'Year': year,
        'Yearly_Return': yearly_return,
        'Yearly_Volatility': yearly_volatility,
        'Max_Drawdown': max_drawdown
    })

# Create DataFrame from yearly data
yearly_analysis = pd.DataFrame(yearly_data)

print("\nYearly Analysis (Head):")
print(yearly_analysis.head())

print("\nYearly Analysis (Tail):")
print(yearly_analysis.tail())
````

## Simulate SIP strategies and calculate performance metrics

### Subtask:
Simulate various SIP strategies and compute their final portfolio value, Sharpe ratio, volatility, and maximum drawdown. The strategies include fixed investment on the best day of the month, every Monday, the historically best weekday, a random day each month, and the first day of the month.


**Reasoning**:
First, I need to define the `calculate_sip_performance` function, which will be used to compute various performance metrics for each SIP strategy based on its cumulative growth series and initial investment amount. This function will calculate total invested capital, final portfolio value, total return, annual return, annual volatility, Sharpe ratio, and maximum drawdown.



````python
import numpy as np

def calculate_sip_performance(cumulative_growth_series, initial_investment_amount=1000):
    if cumulative_growth_series.empty:
        return {
            'Strategy': 'N/A',
            'Total Invested': 0,
            'Final Value': 0,
            'Total Return (%)': 0,
            'Annual Return (%)': 0,
            'Annual Volatility (%)': 0,
            'Sharpe Ratio': 0,
            'Max Drawdown (%)': 0
        }

    # Total invested capital
    total_invested = len(cumulative_growth_series) * initial_investment_amount

    # Final portfolio value
    final_value = cumulative_growth_series.iloc[-1]

    # Total Return
    total_return = ((final_value / total_invested) - 1) * 100 if total_invested > 0 else 0

    # Calculate daily returns from cumulative values
    # The cumulative_growth_series represents the total portfolio value at each investment point.
    # To get daily returns of the portfolio for volatility and Sharpe, we need to treat
    # the last value of each period as the 'close' price for that period.
    # However, the previous `simulate_sip_growth_flexible` function calculates `cumulative_value`
    # which includes the new investment. This is essentially the portfolio value right after
    # making an investment on that specific day.

    # To correctly calculate returns for Sharpe Ratio and volatility, we need to consider the
    # growth of the *existing* portfolio for each period. The `Growth_Factor` in the simulation
    # was applied to the *previous_portfolio_value*.
    # Let's use the underlying 'Next_Day_Return' of the actual investment days for volatility, as it's more direct.
    # However, for the whole portfolio, it's more complex as it's a sum of growing investments.
    # A simpler approach for annual volatility and Sharpe ratio on SIP is to calculate the daily
    # returns of the *portfolio* (not individual stock).

    # For simplicity, let's use the daily returns for the days when investments were made
    # This is not perfectly accurate for portfolio volatility but a common approximation for SIP analysis.
    # Reconstructing the actual portfolio daily returns would require simulating the entire portfolio
    # day by day, not just on investment days.

    # A more robust way: calculate daily returns from the 'Close' prices associated with the investment days.
    # This also isn't quite right, as SIP performance depends on the compounding of each investment.

    # For this task, let's derive portfolio daily returns from the `cumulative_growth_series` values.
    # This assumes `cumulative_growth_series` represents the total value of the portfolio at the point of investment.
    portfolio_daily_returns = cumulative_growth_series.pct_change().dropna() * 100

    # Number of trading periods (assuming daily investment opportunities, but SIP is monthly/weekly)
    # The number of years is derived from the date range of the SIP series
    num_years = (cumulative_growth_series.index.max() - cumulative_growth_series.index.min()).days / 365.25

    # Annual Return
    # For SIP, Annual Return can be approximated using CAGR on total invested vs final value
    annual_return = ((final_value / total_invested) ** (1 / num_years) - 1) * 100 if num_years > 0 and total_invested > 0 else 0

    # Annual Volatility (daily volatility * sqrt(252 trading days))
    annual_volatility = portfolio_daily_returns.std() * np.sqrt(252) if not portfolio_daily_returns.empty else 0

    # Sharpe Ratio (assuming risk-free rate is 0 for simplicity)
    # Sharpe Ratio = (Annual Return - Risk-Free Rate) / Annual Volatility
    sharpe_ratio = (annual_return / annual_volatility) if annual_volatility > 0 else 0

    # Max Drawdown
    # Need to calculate Max Drawdown on the cumulative value series
    max_drawdown = 0
    if not cumulative_growth_series.empty:
        peak_value = cumulative_growth_series.iloc[0]
        for val in cumulative_growth_series:
            if val > peak_value:
                peak_value = val
            drawdown = (peak_value - val) / peak_value * 100
            if drawdown > max_drawdown:
                max_drawdown = drawdown

    return {
        'Total Invested': total_invested,
        'Final Value': final_value,
        'Total Return (%)': total_return,
        'Annual Return (%)': annual_return,
        'Annual Volatility (%)': annual_volatility,
        'Sharpe Ratio': sharpe_ratio,
        'Max Drawdown (%)': max_drawdown
    }
````

````python
import pandas as pd

# Create the 'Day_Category' column
def categorize_day_of_month(day):
    if 1 <= day <= 5:
        return 'Start of Month'
    elif 6 <= day <= 25:
        return 'Mid-Month'
    elif 26 <= day <= 31:
        return 'End of Month'
    else:
        return 'Invalid Day'

df['Day_Category'] = df['Day_of_Month'].apply(categorize_day_of_month)

# Group by 'Day_Category' and calculate the average 'Next_Day_Return'
day_category_analysis = df.groupby('Day_Category')['Next_Day_Return'].mean().reset_index()

# Define a specific order for categories for consistent plotting
category_order = ['Start of Month', 'Mid-Month', 'End of Month', 'Invalid Day']
day_category_analysis['Day_Category'] = pd.Categorical(day_category_analysis['Day_Category'], categories=category_order, ordered=True)
day_category_analysis = day_category_analysis.sort_values('Day_Category')

print("Average Next Day Return by Day Category:")
print(day_category_analysis)
````
